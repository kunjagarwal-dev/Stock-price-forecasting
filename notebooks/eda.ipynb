```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv("../data/raw/MSFT.csv", index_col="Date", parse_dates=True)
```

```python
df["Daily_Return"]= df["Close"].pct_change()
```

```python
df["Daily_Return"].describe()
```

```python
plt.plot(df["Daily_Return"])
plt.xlabel("Daily Return")
plt.title("Dailt return")
```

```python
df["MA_30"] = df["Close"].rolling(window=30).mean()
df["MA_200"] = df["Close"].rolling(window=200).mean()
```

```python
df[["Close", "MA_30", "MA_200"]].plot(figsize=(14,6))
```

```python
df["Volatility_30"] = df["Daily_Return"].rolling(window=30).std()
```

```python
plt.plot(df["Volatility_30"])
plt.title("30-Days volatility ")
plt.show()
```
